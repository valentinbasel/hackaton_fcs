# nube de palabras 

Las Wordcloud (nubes de palabras) son herramientas útiles para resumir los conceptos más importantes de un texto, una página web o un libro. Cuanto más presente esté una palabra en el texto considerado, más grande aparecerá en la nube de palabras.

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt



texto = "esto es un texto no muy largo, pero con algunas palabras"
wordcloud = WordCloud(width = 800, height = 800,
                    background_color ='white',
                    min_font_size = 10).generate(texto)                    
plt.figure(figsize = (8, 8), facecolor = None)
plt.imshow(wordcloud)
plt.axis("off")
plt.tight_layout(pad = 0)
plt.show()
```

```python
import numpy as np
from PIL import Image

mask = np.array(Image.open("nube.png"))
mask[mask == 1] = 255

wordcloud = WordCloud(background_color = "white", 
                      max_words = 50, 
                      mask = mask).generate(texto)
plt.imshow(wordcloud)
plt.axis("off")
plt.show();
```
