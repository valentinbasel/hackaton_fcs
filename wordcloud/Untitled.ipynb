# Mineria de texto

La minería de texto es un campo de la ciencia de datos que se centra en la extracción de información útil y conocimiento a partir de documentos de texto no estructurados. Estos documentos pueden incluir correos electrónicos, publicaciones en redes sociales, artículos de noticias, informes, libros y más. Los procesos de minería de texto implican una serie de pasos que ayudan a convertir el texto en datos estructurados y comprensibles. 
Hay una serie de procedimientos que se se llevan a cabo generalmente en un proceso de mineria de texto, y estos son:


1. Recopilación de datos
2. Preprocesamiento de texto
3. Tokenización
4. Eliminación de palabras vacías
5. Stemming y lematización
6. Análisis de frecuencia de términos
7. Modelado de temas
8. Análisis de sentimiento
9. Extracción de información
10. Visualización y presentación

## y aca un detalle (generado por chatGPT como para que vayan viendo lo que hacen las IA):

1. Recopilación de datos:
   - Este es el primer paso en el proceso de minería de texto, donde se recopilan los documentos de texto de diversas fuentes. Estos documentos pueden estar en formato digital o físico, y se deben almacenar de manera accesible para su posterior análisis.

2. Preprocesamiento de texto:
   - Antes de realizar análisis más avanzados, es necesario limpiar y preprocesar los documentos de texto. Esto implica eliminar caracteres especiales, signos de puntuación, números y palabras vacías (stop words) que no aportan información relevante. También se realiza la tokenización, que divide el texto en palabras o términos individuales.

3. Tokenización:
   - La tokenización implica dividir el texto en unidades más pequeñas, como palabras o términos. Cada uno de estos términos se conoce como un "token" y se utiliza como base para el análisis posterior.

4. Eliminación de palabras vacías:
   - Las palabras vacías, como "el", "de", "y", que aparecen con frecuencia en el texto pero no aportan un significado sustancial, se eliminan para reducir el ruido en el análisis.

5. Stemming y lematización:
   - Estos procesos se utilizan para reducir las palabras a su forma base. El stemming busca eliminar sufijos y prefijos, mientras que la lematización busca reducir las palabras a su forma léxica básica. Esto ayuda a agrupar palabras similares y simplifica el análisis.

6. Análisis de frecuencia de términos:
   - Se calcula la frecuencia de cada término en el corpus (conjunto de documentos) para determinar cuán importante es en el contexto de los documentos. Esto puede ayudar a identificar palabras clave y temas relevantes.

7. Modelado de temas:
   - Se utilizan técnicas como la modelización de temas, que identifica patrones y temas clave en el texto. Los modelos de temas pueden ayudar a agrupar documentos relacionados y descubrir tendencias ocultas.

8. Análisis de sentimiento:
   - Se puede realizar un análisis de sentimiento para determinar la actitud o emoción expresada en el texto. Esto es útil en aplicaciones como la monitorización de redes sociales y la opinión del cliente.

9. Extracción de información:
   - En este paso, se extraen entidades y relaciones relevantes del texto, como nombres de personas, lugares, fechas y eventos. Esto puede ser útil en la creación de bases de datos estructuradas a partir de información no estructurada.

10. Visualización y presentación:
    - Finalmente, los resultados del análisis de texto se visualizan de manera comprensible para ayudar a los usuarios a comprender los patrones, tendencias y conocimientos extraídos de los documentos de texto. Esto puede incluir gráficos, tablas y resúmenes.

La minería de texto es una herramienta poderosa para obtener información valiosa a partir de grandes volúmenes de datos no estructurados, y se utiliza en una variedad de campos, como la inteligencia empresarial, la investigación académica, la análisis de redes sociales, la clasificación de documentos y más.

```python
# Recopilación de datos

texto = "soy un texto con mucha información, Tiene MAYUSCULAS que no es lo mismo que mayusculas o mayusCULAS"
```

```python
# Preprocesamiento de texto (normalización)
# la forma "rustica" (pero funciona)
texto = texto.lower() # poner todo en minuscula
texto = texto.replace(",","") # borrar simbolos especiales
print(texto)
```

```python
# tokenización (también, la forma muyyyy rustica)

texto = texto.split(" ")
print(texto)
```

```python
#Eliminación de palabras vacías (stopword), todo muy rustico pero para que vean como funcionan con las lib estandar

palabras_vacias = ["que","es","lo","o","con","un","no"] #las palabras a borrar
texto_limpio = [] # la nueva lista de palabras limpia

for datos in texto:
    if datos not in palabras_vacias:
        texto_limpio.append(datos)

print(texto_limpio)
```

```python
# una gran ELIPSIS y saltemos a la ultima etapa:
# Visualización y presentación

from wordcloud import WordCloud
import matplotlib.pyplot as plt

#unimos el texto limpio para pasarlo a wordcloud
texto_final="" #a wordcloud le hace falta una cadena de string no una lista

for txt in texto_limpio:
    texto_final = texto_final + txt + " "

print(texto_final)

# y ahora MAGIA
wordcloud = WordCloud(width = 800, height = 800,
                    background_color ='white',
                    min_font_size = 10).generate(texto_final)

# matplot es para graficar, es una herramienta gigante, pero no es para una sola clase

plt.figure(figsize = (8, 8), facecolor = None)
plt.imshow(wordcloud)
plt.axis("off")
plt.tight_layout(pad = 0)
plt.show()
```
